# Pertemuan 03 — Percabangan dan Validasi


```python
def grade(nilai):
    if nilai < 0 or nilai > 100: return None
    if nilai >= 90: return 'A'
    if nilai >= 80: return 'B'
    if nilai >= 75: return 'C / Lulus'
    return 'Belum lulus'
for n in [74,75,76,89,90,100,-1,101]: print(n, grade(n))
```

```python
def boleh_naik(usia,tinggi): return usia>=10 and tinggi>=130
for u,t in [(9,140),(10,129),(10,130)]: print(u,t,boleh_naik(u,t))
```
