## 1. Sistem Nilai
Ubah `nilai` menjadi 74, 75, 76, 89, 90, -1, 100, dan 101; Run ulang cell dan catat expected-versus-actual.


```python
nilai = 75
if nilai < 0 or nilai > 100:
    print('Nilai tidak valid')
elif nilai >= 90:
    print('A')
elif nilai >= 80:
    print('B')
elif nilai >= 75:
    print('C / Lulus')
else:
    print('Belum lulus')
```

## 2. Dua Syarat dengan AND
Ubah nilai boolean satu per satu untuk melihat seluruh kombinasi.


```python
syarat_a = True
syarat_b = False
if syarat_a and syarat_b:
    print('Kedua syarat terpenuhi')
else:
    print('Belum semua syarat terpenuhi')
```

## 3. OR dan NOT


```python
a = True
b = False
print('A and B:', a and b)
print('A or B :', a or b)
print('not A  :', not a)
```

## 4. Klasifikasi Interval
Uji 17, 18, 27, dan 28.


```python
suhu = 18
if suhu < 18:
    print('Dingin')
elif suhu <= 27:
    print('Nyaman')
else:
    print('Panas')
```
