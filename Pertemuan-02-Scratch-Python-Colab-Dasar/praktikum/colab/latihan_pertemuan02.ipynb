## 1. Output dan Variable


```python
pesan = "Halo, Python!"
angka = 7
print(pesan)
print("Angka:", angka)
```

## 2. Kalkulator Belanja — Input, Conversion, Operator


```python
nama_barang = input("Nama barang: ")
harga = float(input("Harga satuan: "))
jumlah = int(input("Jumlah: "))
total = harga * jumlah
print(f"Total: Rp{total:,.0f}")
```

## 3. Luas dan Keliling Persegi Panjang


```python
sisi_a = float(input("Sisi A: "))
sisi_b = float(input("Sisi B: "))
luas = sisi_a * sisi_b
keliling = 2 * (sisi_a + sisi_b)
print("Luas:", luas)
print("Keliling:", keliling)
```

## 4. Konversi Menit — // dan %


```python
menit = int(input("Total menit: "))
jam = menit // 60
sisa = menit % 60
print(jam, "jam", sisa, "menit")
```

## 5. Konversi Suhu


```python
celsius = float(input("Celsius: "))
fahrenheit = celsius * 9 / 5 + 32
kelvin = celsius + 273.15
print("Fahrenheit:", fahrenheit)
print("Kelvin:", kelvin)
```

## 6. Estimasi Perjalanan — Sequence Saja
Gunakan nilai input valid. Validasi condition dibahas pada Pertemuan 03.


```python
jarak_km = float(input("Jarak km: "))
km_per_liter = float(input("Efisiensi km/liter: "))
nilai_per_liter = float(input("Nilai per liter: "))
liter = jarak_km / km_per_liter
total_nilai = liter * nilai_per_liter
print(f"Kebutuhan: {liter:.2f} liter")
print(f"Estimasi total: {total_nilai:,.0f}")
```
