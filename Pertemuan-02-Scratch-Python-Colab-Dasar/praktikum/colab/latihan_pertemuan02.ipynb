## Biodata


```python
nama = input("Nama: ").strip()
umur = int(input("Umur: "))
kota = input("Kota: ").strip()
print(f"Halo {nama}! Umur {umur}, dari {kota}.")
```

## Kalkulator Belanja


```python
nama_barang = input("Nama barang: ")
harga = float(input("Harga satuan: "))
jumlah = int(input("Jumlah: "))
total = harga * jumlah
print(f"Total: Rp{total:,.0f}")
```

## Konversi Menit


```python
menit = int(input("Total menit: "))
print(menit // 60, "jam", menit % 60, "menit")
```

## Konversi Suhu


```python
c = float(input("Celsius: "))
print("Fahrenheit:", c*9/5+32)
print("Kelvin:", c+273.15)
```

## Estimasi Perjalanan


```python
jarak=float(input("Jarak km: "))
ef=float(input("km/l: "))
harga=float(input("Harga/l: "))
if ef <= 0: print("Efisiensi harus > 0")
else:
    liter=jarak/ef
    print(f"BBM {liter:.2f} L, biaya Rp{liter*harga:,.0f}")
```
