# Pertemuan 04 — Loop, Counter, Akumulasi


```python
saldo=0
for i in range(1,4):
    saldo += 10000
    print(i,saldo)
```

```python
for n in range(1,21):
    if n % 2 == 0: print(n,end=' ')
```

```python
for baris in range(1,6):
    print('*'*baris)
```
