## 1. For + Accumulator


```python
total = 0
for i in range(1, 4):
    total += 10000
    print(i, total)
```

## 2. Range + Bilangan Genap


```python
for n in range(1, 21):
    if n % 2 == 0:
        print(n, end=' ')
```

## 3. Lima Nilai dan Rata-rata Tanpa List


```python
total_nilai = 0
for i in range(1, 6):
    nilai = float(input(f'Nilai ke-{i}: '))
    total_nilai += nilai
rata_rata = total_nilai / 5
print('Total:', total_nilai)
print('Rata-rata:', rata_rata)
```

## 4. While sampai Condition Terpenuhi


```python
target = 3
nilai = 0
while nilai != target:
    nilai = int(input('Masukkan angka target 3: '))
    print('Nilai saat ini:', nilai)
print('Condition selesai')
```

## 5. Sentinel 0


```python
total_data = 0
nilai = float(input('Nilai, 0 untuk selesai: '))
while nilai != 0:
    total_data += nilai
    nilai = float(input('Nilai, 0 untuk selesai: '))
print('Total:', total_data)
```

## 6. Nested Loop


```python
for baris in range(1, 5):
    for kolom in range(baris):
        print('*', end='')
    print()
```
