# Pertemuan 06 — Function, Dekomposisi, Debugging


```python
def hitung_total(harga,jumlah): return harga*jumlah
def format_rupiah(n): return f'Rp{n:,.0f}'
print(format_rupiah(hitung_total(10000,3)))
```

```python
def rata_rata(data):
    if not data: raise ValueError('data kosong')
    return sum(data)/len(data)
for d in ([80,90,70],[75],[]):
    try: print(d,rata_rata(d))
    except ValueError as e: print(d,e)
```
